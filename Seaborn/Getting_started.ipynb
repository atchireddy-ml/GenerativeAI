#### 2.Distrubution Plots ---> These plots describe the distrubution of a numerical column in  a dataset such the what kind of distrubution normal or skewed.Also outlers can be determined and important part is central tendency and spread can be seen more precisely 
### Types of Distrubution Plots 
## (i)Histogram--> It is a type of distrubution plot in which the distrubution is found out through bins which equally divides the dataset into equal intervals
#### params--> x,element,hue,data,bins
#### Figure level Plot --> displot() 
## (ii)kdeplot --> It is a plot in which instead of bins a smooth curve (like a probability distrubution function) representing the distrubution of a column 

```python
sns.histplot(data=df,x='Age',bins=20)
```

```python
gap.head()
```

```python
sns.histplot(data=gap,x='lifeExp',kde=True)
```

```python
tips.head()
```

```python
sns.histplot(data=tips,x='total_bill',bins=20,hue='sex',element='step')
```

```python
sns.displot(data=tips,x='total_bill',bins=20,col='sex',kind='hist')
plt.show()
```

```python
sns.kdeplot(data=tips,x='total_bill',hue='sex',fill=True)
plt.show()
```

```python
sns.displot(data=gap,x='lifeExp',kind='kde')
plt.show()
```

#### 3.rugplot ---> This plot adds small vertical ticks along the axis where more ticks indicates more density

#### Matrix plot ---> In This type of plots both dimensions are represented as a table called 2d matrix,through these kind of plots correlation,similarity between the columns can be found out 
#### The dataframe is First converted into pivot table before turning it into a matrix (params--> index,columns,values)
#### HeatMap --> In the matrix the values are represented with color indices such that based on the colors the relation about the columns can be betterly understood 
#### params of a heatmap:
###  (i)data(ii)annot-->provides the numerical values if wanted(iii)cmap-->The styling or theme of the heatmap(iv)linewidth-->proper spacing between the lines of a column in the matrix 

```python
gap.head()
```

```python
gap['continent'].unique()
```

```python
##pivot table
dt=gap[gap['continent']=='Europe'].pivot(index='country',columns='year',values='lifeExp')
```

```python
##HeatMap
plt.figure(figsize=(15,15))
sns.heatmap(data=dt,annot=True,cmap='viridis')
plt.show()
```

#### Multiplots --> joint plot and pair plot are two kinds of multiplots 
## pair plot --> This plot shows all the  distrubutions of numerical columns in the dataset 
## Off-diagonal cells: Display a scatter plot between two different variables to expose correlations, patterns, and outliers.
## Diagonal cells: Display a univariate distribution plot (like a histogram) showing the range and frequency of that single variable 

```python
sns.pairplot(data=tips)
```

```python
sns.pairplot(data=sales)
```

#### joint plot --> It is used when we want the detailed view of the 2 numerical columns

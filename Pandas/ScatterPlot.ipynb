```python
import pandas as pd
df=pd.read_csv('Chocolate Sales.csv')
df['Amount']=df['Amount'].str.replace("$", "", regex=False).str.replace(",", "", regex=False)
```

```python
df.head()
```

```python
df['Amount']=df['Amount'].astype('int64')
```

```python
df['month'] = pd.to_datetime(df['Date'], format='%d-%b-%y').dt.strftime('%B').str.lower()
```

```python
import matplotlib.pyplot as plt
x=df['month']
y=df['Boxes Shipped']
plt.figure(figsize=(10,4))
plt.title('Total number of Sales over months in a year')
plt.xlabel('Months')
plt.ylabel('Sales')
plt.plot(x,y,color='Red')

```

## Scatter plot with total number of boxes shipped with Amount(Nothing but the correlation between them)

```python
import matplotlib.pyplot as plt
x=df['Boxes Shipped']
y=df['Amount']
plt.figure(figsize=(10,4))
plt.title('Total number of boxes shipped and their corresponding Amount')
plt.xlabel('Number of Boxes')
plt.ylabel('Amount')
plt.scatter(x,y,color='Red',marker='o')

```

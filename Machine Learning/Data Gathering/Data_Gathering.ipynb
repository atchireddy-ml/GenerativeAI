####  Converters are functions which can change the specific value of a dataset 

```python
def station(value):
    if value=='Q':
        return 'QueensTown'
    else:
        return value
```

```python
df=pd.read_csv('Titanic-Dataset.csv',converters={'Embarked':station})
df.head(10)
```

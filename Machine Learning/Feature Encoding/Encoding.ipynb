#### Basic overview of encoding --> OneHotEncoding(nominal data),Ordinal Encoder(order),LabelEncoder 

```python
import pandas as pd
import sklearn 
```

```python
bank=pd.read_csv('Bank_Marketing_Original_Data.csv')
```

#### ordinal data ---> tertiary > secondary > primary > others
#### so encoding will be done as 3 2 1 0 

```python
bank['Education'].value_counts()
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
ordinal_data=bank[['Education']]
oe=OrdinalEncoder(categories=[['unknown','primary','secondary','tertiary']])
```

```python
oe.fit_transform(ordinal_data)
```

#### Target column is often referred as Label,so  Label Encoding should be performed on it 

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
```

```python
Target_Data=bank[['TARGET']]
le.fit_transform(Target_Data)
```

```python
from sklearn.preprocessing import OneHotEncoder
oe = OneHotEncoder()
```

```python
nominal_data=bank[['Job','Marital','Default','Balance','Housing','Loan','contact','Month','poutcome']]
nominal_data=oe.fit_transform(nominal_data)
```

#### Column Transformer -->  In the previous approach we say that we did encoding in a naive way and by spliting the columns and we nned to add them to again reflect in the dataframe and that isa time taking process 
### so we use the column transformer performs the operations on the dataframe without splitting them 
### simple imputer can also used by the means of column transformer   

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OrdinalEncoder

 
```

```python
col_trans=ColumnTransformer(transformers=[('t1',OrdinalEncoder(categories=[['unknown','primary','secondary','tertiary']]),['Education']),('t2',OneHotEncoder(),['Job','Marital','Default','Balance','Housing','Loan','contact','Month','poutcome'])],remainder='passthrough')
```

```python
X=bank.iloc[:,:-1]
```

```python
X
```

```python
Y=bank[['TARGET']]
```

```python
X_Transformed=col_trans.fit_transform(X)
```

```python
le=LabelEncoder()
le.fit_transform(Y)
```

### Part-1 --> Regression Algorithm 

## Task-1 ---> Linear Regression

```python
import pandas as pd
import csv
df = pd.read_csv(
    'household_power_consumption.csv',
    sep=';',
    na_values='?',
    encoding='latin1',
    quoting=csv.QUOTE_NONE
)
# Remove surrounding double quotes
df.columns = df.columns.str.strip('"')
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].str.strip('"')
```

```python
df
```

```python
df['Datetime'] = pd.to_datetime(
    df['Date'] + ' ' + df['Time'],
    format='%d/%m/%Y %H:%M:%S'
)
```

```python
df
```

```python
df['Year'] = df['Datetime'].dt.year
df['Month'] = df['Datetime'].dt.month
df['Day'] = df['Datetime'].dt.day
df['Hour'] = df['Datetime'].dt.hour
df['Minute'] = df['Datetime'].dt.minute
df['DayOfWeek'] = df['Datetime'].dt.dayofweek
df['IsWeekend'] = (df['DayOfWeek'] >= 5).astype(int)
df.drop(columns=['Date', 'Time', 'Datetime'], inplace=True)
```

```python
df
```

```python
numerical_data = [
    'Global_active_power',
    'Global_reactive_power',
    'Voltage',
    'Global_intensity',
    'Sub_metering_1',
    'Sub_metering_2',
    'Sub_metering_3'
]

for col in numerical_data:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
df=df[['Global_active_power','Global_reactive_power','Voltage','Global_intensity','Sub_metering_1','Sub_metering_2','Sub_metering_3','Month','Hour','DayOfWeek','IsWeekend']]
```

```python
df.info()
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
numerical_Pipeline=Pipeline(steps=[('impute',SimpleImputer(strategy='median')),('scaling',StandardScaler())])
numerical_data=['Global_reactive_power','Voltage','Global_intensity','Sub_metering_1','Sub_metering_2','Sub_metering_3','Month','Hour','DayOfWeek','IsWeekend']
features=ColumnTransformer(transformers=[('Numerical_Transformation',numerical_Pipeline,numerical_data)])
df = df.dropna(subset=['Global_active_power'])  
```

```python
X=df.drop(columns=['Global_active_power'])
Y=df['Global_active_power']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2)
```

```python
from sklearn.linear_model import LinearRegression
model=Pipeline(steps=[('features',features),('regressor',LinearRegression())])
model.fit(X_train,Y_train)
y_pred=model.predict(X_test)
```

### Task-2

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np
mae = mean_absolute_error(Y_test, y_pred)

mse = mean_squared_error(Y_test, y_pred)

rmse = np.sqrt(mse)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
```

## Task-7 ----> Overfitting & Underfitting


```python
from sklearn.metrics import r2_score 
train_pred = model.predict(X_train)
test_pred = model.predict(X_test)
print("Training R2:", r2_score(Y_train, train_pred))
print("Testing R2:", r2_score(Y_test, test_pred))
```

```python
from sklearn.tree import DecisionTreeRegressor
complex_model = Pipeline(steps=[
    ('features', features),
    ('regressor', DecisionTreeRegressor(
        max_depth=None,
        min_samples_split=2,
        min_samples_leaf=1,
        random_state=42
    ))
])
complex_model.fit(X_train, Y_train)
train_pred_complex = complex_model.predict(X_train)
test_pred_complex = complex_model.predict(X_test)

print("Training R2:", r2_score(Y_train, train_pred_complex))
print("Testing R2:", r2_score(Y_test, test_pred_complex))
```

```python

comparison = pd.DataFrame({
    'Model': ['Simple Model', 'Complex Model'],
    'Training R2': [
        r2_score(Y_train, train_pred),
        r2_score(Y_train, train_pred_complex)
    ],
    'Testing R2': [
        r2_score(Y_test, test_pred),
        r2_score(Y_test, test_pred_complex)
    ]
})

print(comparison)
```

### Part-3 ----> Classification Algorithms 

## Task-3 --> Logistic Regression 

```python

import sklearn
bank=pd.read_csv('bank-full.csv',sep=';')
```

```python
bank
```

```python
bank=bank[['age','job','education','balance','housing','loan','contact','month','duration','campaign','pdays','previous','poutcome','y']]
```

```python
numerical_data=['age','balance','duration','campaign','pdays','previous']
ordinal_data=['education']
nominal_data=['job','housing','loan','contact','month','poutcome']
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder,OrdinalEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.compose import ColumnTransformer 
```

```python
numerical_Pipeline=Pipeline(steps=[('impute',SimpleImputer(strategy='median')),('scaling',StandardScaler())])
ordinal_Pipeline=Pipeline(steps=[('impute',SimpleImputer(strategy='most_frequent')),('encoding',OrdinalEncoder(categories=[['unknown','primary','secondary','tertiary']])),('scaling',StandardScaler())])
nominal_Pipeline=Pipeline(steps=[('impute',SimpleImputer(strategy='most_frequent')),('encoding',OneHotEncoder(handle_unknown='ignore'))])
```

```python
features=ColumnTransformer(transformers=[('Numerical_Transformation',numerical_Pipeline,numerical_data),('Ordinal_Transformation',ordinal_Pipeline,ordinal_data),('Nominal_Transformation',nominal_Pipeline,nominal_data)])
```

```python
features
```

```python
from sklearn.linear_model import LogisticRegression
model = Pipeline(steps=[('features',features),('classifier',LogisticRegression())])
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
Y=bank['y']
Y=le.fit_transform(Y)
X=bank.iloc[:,:-1]
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2)
model.fit(X_train,Y_train)
y_pred=model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_pred,Y_test)
```

## Task-4 ---> GaussianNB 

```python
from sklearn.naive_bayes import GaussianNB
model1=Pipeline(steps=[('features',features),('classifier',GaussianNB())])
model1.fit(X_train,Y_train)
y_pred=model1.predict(X_test)
accuracy_score(y_pred,Y_test)
```

## Task-5 ----> K-Nearest Neighbors (KNN)

```python
from sklearn.neighbors import KNeighborsClassifier
model2=Pipeline(steps=[('features',features),('classifier',KNeighborsClassifier(n_neighbors=17))])
model2.fit(X_train,Y_train)
y_pred=model2.predict(X_test)
accuracy_score(y_pred,Y_test)
```

## Task-6 ----> Classification Metrics 

```python
from sklearn.metrics import precision_score, recall_score, f1_score
model2.fit(X_train,Y_train)
y_pred=model2.predict(X_test)
print(precision_score(y_pred,Y_test))
print(recall_score(y_pred,Y_test))
print(f1_score(y_pred,Y_test))
```

```python
model1.fit(X_train,Y_train)
y_pred=model1.predict(X_test)
print(precision_score(y_pred,Y_test))
print(recall_score(y_pred,Y_test))
print(f1_score(y_pred,Y_test))
```

```python
model.fit(X_train,Y_train)
y_pred=model.predict(X_test)
print(precision_score(y_pred,Y_test))
print(recall_score(y_pred,Y_test))
print(f1_score(y_pred,Y_test))
```
